```python
import pandas as pd
import requests
```

```python
df=pd.DataFrame()
for i in range(1,363):
    response=requests.get('https://api.themoviedb.org/3/movie/now_playing?api_key=37fb2a0ccbe5973006934030678e872b&language=en-US&page={}'.format(i))
    temp_df=pd.DataFrame(response.json()['results'])[['id','title','overview','popularity','release_date','vote_average','vote_count']]
    df=pd.concat([df,temp_df],ignore_index=False)

```
