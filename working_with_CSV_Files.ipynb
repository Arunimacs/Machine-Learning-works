**2. Opening a csv file from an URL**

```python
import requests
from io import StringIO

url = "https://raw.githubusercontent.com/Arunimacs/Machine-Learning-works/refs/heads/main/countries.csv"
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url, headers=headers)
data = StringIO(req.text)

pd.read_csv(data)
```

**3. Sep Parameter**

```python
#working with tsv files.tsv files has tab separating the columns.
pd.read_csv('/content/movie_titles_metadata.tsv')
```

```python
#using sep paramer to seperate the columns
pd.read_csv('/content/movie_titles_metadata.tsv',sep='\t')
```

```python
#use name parameter to provide column headings.otherwise the first row is treated as the columns headings
pd.read_csv('/content/movie_titles_metadata.tsv',sep='\t',names=['sno','name','release_year','rating','votes','genres'])
```

**13. converters**

```python
def rename(name):
  if name=='Royal Challengers Bangalore':
    return 'RCB'
  else:
    return name
```

```python
pd.read_csv('IPL Matches 2008-2020.csv',converters={'team1':rename})
```
