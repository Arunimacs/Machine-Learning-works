```python
df=pd.DataFrame()
for i in range(1,501):
    response=requests.get('https://api.themoviedb.org/3/movie/top_rated?api_key=37fb2a0ccbe5973006934030678e872b&language=en-US&page={}'.format(i))
    temp_df=pd.DataFrame(response.json()['results'])[['id','title','release_date','overview','popularity','vote_average','vote_count']]
    df=pd.concat([df,temp_df],ignore_index=True)
```

```python
df
```

```python
df.to_csv('movies.csv',index=False)
```
